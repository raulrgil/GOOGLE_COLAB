```python
# prompt: crea un grafico de barras con datos de ejemplo con distintos colores

import matplotlib.pyplot as plt
import random

# Datos de ejemplo
categorias = ['A', 'B', 'C', 'D', 'E']
valores = [random.randint(1, 20) for _ in range(5)]
colores = ['red', 'green', 'blue', 'yellow', 'purple']

# Crear el gráfico de barras
plt.figure(figsize=(8, 6))  # Ajustar tamaño de la figura
plt.bar(categorias, valores, color=colores)

# Agregar etiquetas y título
plt.xlabel('Categorías')
plt.ylabel('Valores')
plt.title('Gráfico de Barras con Colores')

# Mostrar el gráfico
plt.show()
```
